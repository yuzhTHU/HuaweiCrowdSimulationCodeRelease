```python
import pandas as pd
from tqdm import tqdm
from pathlib import Path

path_list = [
    "logs/train/20251028_Baseline_223639_DL4",
    "logs/train/20251028_BS64_223645_DL4",
    "logs/train/20251029_BS128_115642_DL4",
    "logs/train/20251028_denoise2_223825_DL4",
    "logs/train/20251029_denoise5_115628_DL4",
    "logs/train/20251029_lr2e-4_115638_DL4",
    "logs/train/20251028_lr5e-3_223706_DL4",
    "logs/train/20251028_sample10_223720_DL4",
    "logs/train/20251029_sample5_115632_DL4",
    "logs/train/20251029_sample1_102302_DL3",
    "logs/train/20251029_dim256_120400_DL3",
    "logs/train/20251029_dim64_120353_DL3",
    "logs/train/20251028_scale10_223849_DL4",
    "logs/train/20251028_scale0.1_223855_DL4",
    "logs/train/20251029_LSTM1_120325_DL3",
    "logs/train/20251029_LSTM2_120339_DL3",
    "logs/train/20251029_drop0.5_120344_DL3",
    "logs/train/20251029_drop0.1_120304_DL3",
    "logs/train/20251028_offset10_223842_DL4",
    "logs/train/20251029_predict-acc_120408_DL3",
    "logs/train/20251028_no_antithetic_223900_DL4",
]

df_list = []
for path in tqdm(path_list):
    path = Path(path)
    if not (path / "records.jsonl").exists():
        print(f"Skipping {path}/records.jsonl not found.")
        continue
    df = pd.read_json(path / "records.jsonl", lines=True)
    df['path'] = str(path)
    df['name'] = path.name.split('_', 1)[1].rsplit('_', 2)[0]
    df_list.append(df)
df_all = pd.concat(df_list, ignore_index=True)
df_train = df_all[df_all['ade'].isna()]
df_test = df_all[df_all['ade'].notna()]
df_test
```

```python
df_best = df_test.groupby('name').apply(lambda x: x.loc[x['accuracy'].idxmax()])
df_best
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(data=df_best.reset_index(drop=True).sort_values('accuracy'), y='name', x='accuracy')
plt.xlim(0.85, 0.95)
```

```python
df_list = []
for idx, row in df_best.copy().iterrows():
    is_list = [col for col in row.keys() if isinstance(row[col], list)]
    not_list = [col for col in row.keys() if col not in is_list]
    df = pd.DataFrame(row[is_list].to_dict())
    for col in not_list:
        df[col] = row[col]
    df_list.append(df)
df_detail = pd.concat(df_list, ignore_index=True)
df_detail
```

```python
df_plot = []
for key, group in df_detail.groupby(['name', 'dataset_class']):
    accuracy = 1 - group['ade'] / group['trajlen']
    df_plot.append({
        'name': key[0],
        'dataset_class': key[1],
        'accuracy': accuracy.mean()
    })
df_plot = pd.DataFrame(df_plot)
sns.barplot(data=df_plot.sort_values('accuracy'), x='name', y='accuracy', hue='dataset_class')
plt.ylim(0.85, 0.95)
plt.gca().set_xticklabels(plt.gca().get_xticklabels(), rotation=90, va='top')
```
