```python
import matplotlib.pyplot as plt
from argparse import Namespace
from src.utils.logger import init_logger
from src.utils.plot import get_fig

init_logger('src')
args = Namespace(
    cache_dataset=True, # False,
    fps=2.5,
    dot_per_meter=5,
    hist_step=8,
    pred_step=1,
    roll_step=12,
    skip_step=1,
)
```

```python
from src.dataset import SDDDataset
args.cache_dataset = True
dataset = SDDDataset.load_data(args, 'data/SDD/annotations/bookstore/video1/annotations.txt')
fi, fig, axes = get_fig(1, 1, AW=6, AH=6, dpi=300)
ax = axes[0]
ax.imshow(dataset.map_data.map, extent=(dataset.map_data.xmin, dataset.map_data.xmax, dataset.map_data.ymin, dataset.map_data.ymax), origin='lower', cmap='gray_r')
ax.set_facecolor('#b2bec3')
for pid, group in dataset.df_data.groupby('id'):
    cmap = plt.cm.get_cmap('viridis')
    ax.plot(group['x'], group['y'], color=cmap(group['f'].max() / dataset.df_data['f'].max()))
```
