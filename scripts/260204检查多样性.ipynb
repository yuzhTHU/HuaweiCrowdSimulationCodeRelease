```python
import json
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
from argparse import Namespace
from src.utils.plot import get_fig

df_list = []
loader = Path('logs/test').glob('eth_single_K=*')
# loader = [Path('logs/test/eth_single_K=2')]
for path in sorted(loader, key=lambda x: x.stem):
    args = Namespace(**json.load(open(path / 'args.json', 'r')))
    k = args.sample_num
    record = pd.read_json(path / 'records.jsonl', lines=True).tail(1)

    df_list.append(record.assign(sample_num=k))

df_results_eth = pd.concat(df_list, ignore_index=True)
df_results_eth
```

```python
df_list = []
loader = Path('logs/test').glob('SDD_single_K=*/records.jsonl')
# loader = [Path('logs/test/SDD_single_K=2')]
for path in sorted(loader, key=lambda x: x.parent.stem):
    args = Namespace(**json.load(open(path.parent / 'args.json', 'r')))
    k = args.sample_num
    record = pd.read_json(path, lines=True).tail(1)

    df_list.append(record.assign(sample_num=k))

df_results_sdd = pd.concat(df_list, ignore_index=True)
df_results_sdd
```

```python
fi, fig, axes = get_fig(1, 2, FW=8, A_ratio=2.0, HS=0.5, sharex=True, sharey=True)

ax = axes[0]
df_plot = pd.DataFrame(dict(
    ade=df_results_eth['ade'].apply(np.mean),
    fde=df_results_eth['fde'].apply(np.mean),
    apd=df_results_eth['apd'].apply(lambda x: np.mean(x) if x[0] is not None else 0.0),
    k=df_results_eth['sample_num'],
))
df_plot = df_plot.sort_values('k')
ax.plot(df_plot['k'], df_plot['ade'], marker='o', label='ADE', markersize=2)
ax.plot(df_plot['k'], df_plot['fde'], marker='o', label='FDE', markersize=2)
ax.plot(df_plot['k'], df_plot['apd'], marker='s', label='APD', color='C2', markersize=2)
ax.set_ylim(-0.1, 1.9)
ax.legend(ncols=1, frameon=False, fontsize=0.7*fi['fontsize'], loc='upper right')
ax.set_xlabel('Sample Num $K$')
ax.set_ylabel('Error / Diversity (m)')
ax.set_title('ETH')

ax = axes[1]
df_plot = pd.DataFrame(dict(
    ade=df_results_sdd['ade'].apply(np.mean),
    fde=df_results_sdd['fde'].apply(np.mean),
    apd=df_results_sdd['apd'].apply(lambda x: np.mean(x) if x[0] is not None else 0.0),
    k=df_results_sdd['sample_num'],
))
df_plot = df_plot.sort_values('k')
ax.plot(df_plot['k'], df_plot['ade'], marker='o', label='ADE', markersize=2)
ax.plot(df_plot['k'], df_plot['fde'], marker='o', label='FDE', markersize=2)
ax.plot(df_plot['k'], df_plot['apd'], marker='s', label='APD', color='C2', markersize=2)
ax.set_ylim(-0.1, 2.2)
ax.set_xlabel('Sample Num $K$')
# ax.set_ylabel('Error (m) / Diversity (m)')
ax.set_title('SDD')
ax.tick_params(axis='y', which='both', length=2)

fig.savefig('figures/vary_sample_num.png', dpi=300, transparent=True, bbox_inches='tight')
fig.savefig('figures/vary_sample_num.pdf', dpi=300, transparent=True, bbox_inches='tight')
fig.savefig('figures/vary_sample_num.svg', dpi=300, transparent=True, bbox_inches='tight')
```
